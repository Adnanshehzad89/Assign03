### Question 7: ATM Menu
Create a simple ATM system using a while loop.

The initial balance is:

50000

Display this menu repeatedly:

1. Check Balance

2. Deposit

3. Withdraw

4. Exit

The program should continue until the user selects 4.

Implement the following:

• Check balance

• Deposit money

• Withdraw money

• Prevent withdrawal if the amount is greater than the balance

• Exit the program 

```python
Balance=50000
while True:
    print("1. Check Balance")
    print("2. Deposite")
    print("3. Withdraw")
    print("4. Exit")
    Num=int(input("Select any One Option: "))
    if Num==4:
        print("Exit")
        break
    elif Num==1:
        print(f"Your Balance is {Balance}")
    elif Num==2:
        print("Deposite Money")
    elif Num==3:
        amount=int(input("Enter the withdraw Amount: "))
        if amount>Balance:
            print("You have Insuficiant Balance")
        else:
            print(f"you withdraw the Amount {amount} your remaining Balance is {Balance-amount}")
    else:
        print("Invalid Option")
        continue
```

### Question 9: Student Marks Entry System

A teacher wants to enter student marks continuously.

Use a while loop.

The program should keep asking for marks until the teacher enters:

-1

For every student's marks:

• 50 or above → Pass

• Below 50 → Fail

At the end, display:

Total Students

Total Passed

Total Failed

The -1 should not be counted as a student's marks.

```python
Total_Students=0
Total_Passed=0
Total_Failed=0


while True:
    Total_Students+=1
    marks = int(input("Enter the Marks of Students: "))
    if marks == -1:
        print(f"Total Students{Total_Students}")
        print(f"Total Passed{Total_Passed}")
        print(f"Total Failed{Total_Failed}")
        break

    elif marks>= 50:
        Total_Passed+=1
        print("Passed")
    else:
        Total_Failed+=1
        print("Failed")
    continue
```
